## Aufgabe 1:
- Liste aller Namen einlesen, der Benutzer soll sie einzeln eingeben und mit [ENTER] Taste bestätigen
- Die Liste wird duch eine leere Eingabe beendet ([ENTER] ohne etwas einzutippen)
- Danach soll das Programm die Liste sortieren und sortiert ausgeben

````python
# Lösung Aufgabe 1

Namen = []
while True:
    AktuellerName = input("Bitte einen Namen eingeben:")
    if AktuellerName != "":
        Namen.append(AktuellerName)
    else:
        break
Namen.sort()
print(Namen)
````

## Aufgabe 2:
Die Liste soll "schön" ausgegeben werden, also
```
1: Conradin
2: Elisa
```

````python
# Lösung Aufgabe 2
# - Variante 1, Vorschlag nach Elisa
for i in range(len(Namen)):
    print(str(i+1)+": "+Namen[i])
````

````python
# So kann man aber auch eine Liste durchiterieren, aber ohne Nummern...:
for i in Namen:
    print(i)
````

````python
# Lösung Aufgabe 2
# - Variante 2, mit `enumerate`

for i,name in enumerate(Namen):
    print(i+1,name)
````

# Und nun bauen wir uns dafür eine Funktion:

````python
# def liste_einlesen
# def listeEinlesen

def listeEinlesen(überschrift,aufforderung):
    eingeleseneElemente = []
    print(überschrift)
    while True:
        neuesElement = input(aufforderung)
        if neuesElement != "":
            eingeleseneElemente.append(neuesElement)
        else:
            break
    return(eingeleseneElemente)

Namen = listeEinlesen("Wir lesen die Klassenliste ein","Bitte einen Namen eingeben, [ENTER] um zu beenden:")
print(Namen)
````
