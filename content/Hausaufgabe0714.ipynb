# Aufgabe 1: Die Zeitplanung
Ein Wanderer möchte wissen, wie schnell wäre er an einer anderen Straßenecke, wenn er einfach quer über alles laufen könnte -- was er nicht kann, weil es ja in der Stadt auch so was außergewöhnliches wie Häuser gibt. 

Die Stadt, wo der Wanderer sich gerade befindet, hat einen rechteckigen Straßenplan, das heißt, es gibt nur Straßen, die Nord-Süd verlaufen und Straßen, die Ost-West verlaufen. Der Wanderer kann nur entlang den Straßen laufen.

Er startet an der Kreuzung A und will zu der Kreuzung B, dafür läuft er zuerst eine Ost-West Straße "X" und dann eine Nord-Süd Straße "Y".

Er zählt dabei seine Schritte entlang jeder der beiden Straßen.

Schreibe ein Programm, der:

1. Fragt, wie schnell der Wanderer geht (Einheit: Schritte pro Minute)
2. Fragt, wie weit er entlang der Straße X gelaufen ist (Einheit: Schritte)
3. Fragt, wie weit er entlang der Straße Y gelaufen ist (Einheit: Schritte)
4. Berechnet, wie weit er gelaufen ist und wie weit der Weg wäre, wenn er direkt von A zu B laufen könnte ("quer Beet")
5. Berechnet, wie lange der Wanderer für seine tatsächlich gelaufene Route brauchte und wie lange er brauchen würde, könnte er direkt laufen.

```

Antwort:

Tatsächlich gelaufen: 140 Minuten. Kürzester Weg: 100 Minuten.
```
