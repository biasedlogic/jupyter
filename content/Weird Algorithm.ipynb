# Das komische Algorithmus

Betrachten Sie einen Algorithmus, der eine positive ganze Zahl als Eingabe annimmt: *N*. Wenn *N* ist gerade, der Algorithmus teilt es durch zwei, und wenn *N* ist ungerade, der Algorithmus multipliziert es mit drei und addiert eins. Der Algorithmus wiederholt dies mit dem Ergebnis des vorherigen Schrittes, bis *N*=1. 

Zum Beispiel die Sequenz für *N* = 3 ist wie folgt:

`3 → 10 → 5 → 16 → 8 → 4 → 2 → 1`

Eure Aufgabe ist es, die Ausführung des Algorithmus für einen bestimmten Wert von *N* zu berechnen und die entsprechende Zahlenfolge auszugeben.

# Eingabe

Die einzige Eingabezeile enthält eine ganze Zahl *N*.

# Ausgabe

Eine Zeile, die alle Werte von *N* die das Algorithmus "durchgeht" beinhaltet.

## Einschränkungen

1 ≤ N ≤ 1'000'000

# Beispiel

## Eingabe:

`3`

## Ausgabe:

`3 10 5 16 8 4 2 1`
