# Die ersten Bewegungen

Das Programm kann nicht hier ausgeführt werden, der Code muss in PyBricks ausgeführt werden (d.h. man muss ihn auf https://beta.pybricks.com eingeben und auf einem LEGO Hub ausführen).

Wer allerdings einen - egal welchen - LEGO Technic oder LEGO City Hub hat, kann den auch auf PyBricks flashen und wenn man dazu auch zwei Technic Motoren hat, kann man den Code dann ausführen, auch wenn man kein LEGO Spike Set hat.

```python
from pybricks.hubs import PrimeHub
from pybricks.pupdevices import Motor, ColorSensor, UltrasonicSensor, ForceSensor
from pybricks.parameters import Button, Color, Direction, Port, Side, Stop
from pybricks.robotics import DriveBase
from pybricks.tools import wait, StopWatch

hub = PrimeHub()

motorLinks  = Motor(Port.A, Direction.COUNTERCLOCKWISE)
motorRechts = Motor(Port.B, Direction.CLOCKWISE)

robot = DriveBase(motorLinks, motorRechts, 54.5, 112)

# Geradeausfahren 100mm = 10cm
robot.straight(100)

# manchmal hilfreich für genaue Drehungen:
robot.use_gyro(True)

# Das stoßt die Bewegung an, wartet aber nicht, bis sie zu Ende ist:
robot.turn( 360,wait=False)
# Das stoßt die Bewegung an, aber wartet, bis sie zu Ende ist
robot.turn(-360,wait)
# Wenn man das ausführt, hat man den Eindruck, die Rechstdrehung wurde ignoriert
# - da wir nicht warten, bis die Bewegung komplett ist, wird der nächste Befehl
# mit der Linksdrehung sofort ausgeführt und die Linksdrehung sofort unterbrochen

# jetzt nochmal wozu das nützlich sein kann:
robot.turn( 360,wait=False)
wait(100)
print("Der Roboter dreht sich schon, während diese Ausgabe angezeigt wird.")
while not robot.done(): # um zu prüfen, ob die letzte Bewegung durch ist, 
    continue            # nutzt man .done(), das gibt True zurück, wenn der/die
                        # Motor/en seine/ihre letzte Anweisung bis Ende ausgeführt
                        # haben
robot.turn(-360,wait)
```
