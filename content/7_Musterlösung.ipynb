# Aufgabe 1

Aufgabe: Verbessert das Programm zum Verdoppeln von Werte so, dass ganze Zahlen auch ganze Zahlen bleiben (dürfen), sprich, dass wenn man 4 eingibt, man eine 8 bekommt, aber wenn man 4.0 eingibt, eine 8.0 rauskommt.

*Hinweis 1:* Man braucht dafür keine neuen Befehle.

*Hinweis 2:* so, wie z.B. ein `if`-Block einen anderen `if`-Block beinhalten kann (oder eine `while` Schleife in einer anderen `while` Schleife sitzen kann), kann auch ein `try` Block andere `try`-Blöcke in sich enthalten.

## Musterlösung

```python
userEntry = input("Gib etwas ein. Wenn es eine Zahl ist, wird sie verdoppelt")
try:
    number = int(userEntry)
    print("Das doppelte von",number,"ist",number*2)
except ValueError:
    try:
        number = float(userEntry)
        print("Das doppelte von",number,"ist",number*2)
    except ValueError:
        print("Das war wohl keine Zahl, die ich verdoppeln könnte")
```

# Aufgabe 2

Lese in eine Liste alle Noten einen Schülers ein. Ähnlich, wie bei den Namenslisten (Wink, Wink...), soll eine leere Eingabe das Einlesen beenden.
Berechne dann den Notenschnitt aus allen Noten.

Das Programm soll erkennen, falls man anstatt einer Note irgendein Schmarrn eingibt und einen Hinweis an den Nutzer ausgeben.
Am Besten soll der Hinweis direkt nach der fehlerhaften Eingabe kommen und die Eingabe soll dann wiederholt werden. Okay ist aber auch, wenn der Hinweis erst beim Berechnen des Notendurchschnitts kommt. Das Programm darf nur nicht mit einem Fehler unterbrochen werden, falls die Eingaben falsch sind.

## Musterlösung

```python
noten = []
while True:
    userInput = input("Gebe die nächste Note ein, [ENTER] beendet:")
    if userInput == "":
        break
    try:
        note = float(userInput)
        noten.append(note)
    except ValueError:
        print("Nur Noten in Form von Zahlen werden unterstützt")
if len(noten)>0:
    print("Notenschnitt liegt bei:",sum(noten)/len(noten))
else:
    print("Man braucht Noten um ein Notenschnitt zu berechnen.")
```

# Aufgabe 3 (extra Aufgabe)

Euer Programm bekommt als Eingabe zwei Zeilen. Die erste Zeile enthält eine Zahl N.

Die zweite Zeile enthält (N-1) verschiedene Zahlen so, dass keine Zahl größer als N ist und keine kleiner als 1.

Das heißt im Klartext, eine Zahl X (1≤X≤N) fehlt, ansonsten sind in der zweiten Zeile alle Zahlen, zufällige Reihenfolge, da.

Es gilt diese Zahl zu finden und auszugeben.

Das Programm muss für 2 ≤ N ≤ 200'000 funktionieren.

## Musterlösung -- nur mit den Infos aus dem Kurs

```python
n = input()
s = input()

### Dieser Teil schneidet die eingelesene Zeichenkette `s` an den Leerzeichen ###
z = []
i = j = 0
while i<len(s):
    while j<len(s) and s[j]!=" ":
        j+=1
    z.append(int(s[i:j]))
    while j<len(s) and s[j]==" ": 
        j+=1
    i=j
##############################       bis hier      ##############################

z.sort()
i = 1
for k in z:
    if k!=i:
        break
    i+=1
print(i)
```

## Musterlösung mit einem neuen Befehl: `str.split()`

`str.split()` teilt eine Zeichenkette in eine Liste an Leerzeichen:

```python
s = "Quick brown fox jumped over the lazy dog.".split()
print(s)
```

Und hier kommt die Lösung:

```python
n = input()
s = input().split()
z = []
for k in s:
    z.append(int(k))
z.sort()
i = 1
for k in z:
    if k!=i:
        break
    i+=1
print(i)
```

## Turbolösung -- ähnlich, aber mit einem kleinen Extra:
Einige Elemente in dieser Lösung wurden (noch) nicht im Kurs erklärt, wer die Notation in der Zeile 2 nachvollziehen möchte, kann gerne hier darüber (auf Deutsch) nachlesen: [Python List Comprehension](https://www.datacamp.com/de/tutorial/python-list-comprehension).

```python
n = int(input())
z = [int(x) for x in input().split()]
z.sort()
for i,k in enumerate(z):
    if k!=i+1:
        print(i+1)
        break
else:
    print(n)
```
