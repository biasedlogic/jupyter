## Hinweis 2

Die Methode `.index(s)` findet, an welchen Stelle s in einer Liste, oder in einer Zeichenkette ist. Achtung, die Methode verursacht einen Fehler, wenn `s` nicht gefunden wurde, siehe Hinweis 3.


````python
text = "abcdef"
print("'abcdef'.index('e') gibt",text.index('e'),"zurück.") 

print("'abcdef'.index('z') gibt",text.index('z'),"zurück.") 
        # hier kommt ein ValueError! 'z' ist nicht im text
````

## Hinweis 3

Der Befehl `in` prüft, ob etwas in einer Liste, oder in einer Zeichenkette vorkommt.  

````python
text = "abcdef"
if 'e' in text:
    print("'abcdef'.index('e') gibt",text.index('e'),"zurück.") 
else:
    print("'e' kommt in 'text' nicht vor")
if 'z' in text:
    print("'abcdef'.index('z') gibt",text.index('z'),"zurück.") 
else:
    print("'z' kommt in 'text' nicht vor")
````
