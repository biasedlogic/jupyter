# Aufgabe Notenschlüssel #1

Schreibe eine Funktion namens `notenSchluessel`, die als Argumente zwei Zahlen nimmt: die Anzahl der im Test gesammelten Punkte und die Anzahl der maximal möglichen Punkte. Am Ende wir das Ganze so aufterufen:

```
# hier kommt Eure Funktion

print("Deine Note bei 45/65 Punkten ist:", notenSchluessel(45,65))
print("Deine Note bei 55/65 Punkten ist:", notenSchluessel(55,65))
```

Der Notenschlüssel sieht so aus:
- \>95%: 1
- \>90%: 2
- \>80%: 3
- \>70%: 4
- \>60%: 5
- ≤60%: 6



## Musterlösung Aufgabe 1

````python
# Prozent = 100 * (wieVielHabeIch / wieVielIstMöglich)

def notenSchluessel(erreichtePunkte,maximalePunkte):
    erreichtProzent = 100*(erreichtePunkte/maximalePunkte)
    if erreichtProzent > 95:
        return(1)
    elif erreichtProzent>90:
        return(2)
    elif erreichtProzent>80:
        return(3)
    elif erreichtProzent>70:
        return(4)
    elif erreichtProzent>60:
        return(5)
    else:
        return(6)
        
print("Deine Note bei 45/65 Punkten ist:", notenSchluessel(45,65))
print("Deine Note bei 55/65 Punkten ist:", notenSchluessel(55,65))
````

# Aufgabe Notenschlüssel #2

wie vorher, allerdings jetzt sind die Schwellen in einer Liste hinterlegt und Eure Funktion bekommt einen dritten Parameter: diese Liste. Am Ende wird es folgendermaßen aussehen:



## Musterlösung Aufgabe 2

````python
notenSchluesselMathe = [60, 70, 80, 90, 95]
notenSchluesselEnglisch = [50, 65, 75, 85, 90]

def notenSchluessel(erreichtePunkte, maximalePunkte, notenSchluessel):
    p = 100*(erreichtePunkte/maximalePunkte)
    Note = 6 # Ich setze hier die Note anfänglich auf 6 ####
    i = 0
    while i<len(notenSchluessel) and p>notenSchluessel[i]:
        i+=1
        Note-=1
    return(Note)

p = 60
q = 65

print("Deine Note für",p,"Punkte von",q,"möglichen, nach Matheschlüssel ist:", notenSchluessel(p,q,notenSchluesselMathe))
print("Deine Note für",p,"Punkte von",q,"möglichen, nach Engischschlüssel ist:", notenSchluessel(p,q,notenSchluesselEnglisch))
````

## Und eine Turbolösung für die Aufgabe 2
Wer möchte, sucht sich die Beschreibung der genützten Funktionen selbst zusammen. Das ist noch nicht unseres Niveau, daher nur für Interessierte und Neugierige. 

Die Lösung schauen wir uns vielleicht später irgendwann mal an...

````python
import bisect
notenSchluesselMathe    = [60, 70, 80, 90, 95]
notenSchluesselEnglisch = [50, 65, 75, 85, 90]

def notenSchluessel(erreichtePunkte, maximalePunkte, notenSchluessel):
    return(len(notenSchluessel)+1-bisect.bisect_left(notenSchluessel,100*(erreichtePunkte/maximalePunkte)))

p = 60
q = 65
print("Deine Note nach Matheschlüssel ist:", notenSchluessel(p,q,notenSchluesselMathe))
print("Deine Note nach Engischschlüssel ist:", notenSchluessel(p,q,notenSchluesselEnglisch))
````
