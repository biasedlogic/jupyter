# Hausaufgabe 21.APR.2026

## a) -- Pflichtaufgabe, das muss schon alleine klappen
Schreibe eine Funktion `quaderVolumen`, die den Volumen einen Quaders berechnet (wow...). Die Formel, um den Volumen *V* eines Quaders zu berechnen lautet:

*V* = *a•b•c*

wo *a*, *b* und *c* die Kantenlängen des Quaders sind. Eure Funktion soll die Kantenlängen als Parameter nehmen und wird zum Beispiel so aufgerufen:

`print(quaderVolumen(50,20,40))`

## b) -- Soll-Aufgabe, ein guter Versuch, der aber nicht 100% funzt, zählt als gemacht
Die Funktion aus dem Punkt a) soll erkennen, falls man (eine oder mehrere) negative Kantenlängen eingibt. In einem solchen Fall soll die Funktion die Zahl -1 als Volumen zurückgeben (egal, welche und wie viele der Kantenlängen falsch waren)

## c) -- Kann-Aufgabe, ein Lob, wer es hinkriegt
Wie bei b), aber es soll auch erkannt werden, falls jemand anstelle von einer Zahl, eine Zeichenkette als Kantenlänge übergibt.
