# Aufgabe 3 -- für die interessierten

Schreibe eine Funktion TikTokKlonk(n), die n Sekunden Lang folgendes ausgibt:
- jede Sekunde, abwechselnd, das Wort 'Tik' oder 'Tok'
- jede fünfte Sekunde auch das Wort 'Klonk'

Zum Beispiel:
```
TikTokKlonk(3)

Tik
Tok
Tik
```
und das in 3 Sekunden, jede Zeile nach jeweils einer Sekunde.

Beziehungsweise
```
TikTokKlonk(8)

Tik
Tok
Tik
Tok
Tik
Klonk
Tok
Tik
Tok
```
Dabei das 'Tik' in der fünften Zeile und das danachfolgende 'Klonk' zusammen ausgegeben wurden (in der 5-ten Sekunde) und die gesamte Laufzeit der Funktion bei 8s lag.

## Hinweis 1:
Die Funktion `sleep(s)` aus dem Modul `time` wartet `s` Sekunden.

## Hinweis 2:
Die Funktion `monotonic()`, ebenfalls aus dem Modul `time` gibt die Zeit, in Sekunden, generell ab Start des Programs/des Rechners, die Zeit ist nicht die Uhrzeit, sondern einfach eine (fließkomma-)Zahl, die jede Sekunde um 1 höher ist. 

Für einen anschaulichen Beispiel siehe unten:

````python
# Hinweis / Hilfe zu Aufgabe 3:
from time import sleep, monotonic

print("Wir starten, die Zeit, die monotonic() zurückgibt ist:",monotonic())
sleep(1)
print("Nach sleep(1), die Zeit, die monotonic() zurückgibt ist:",monotonic())
sleep(0.5)
print("Nach sleep(0.5), die Zeit, die monotonic() zurückgibt ist:",monotonic())
````
