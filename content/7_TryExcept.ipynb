# `Try` -- lass Python versuchen...

Manchmal lohnt nicht, alle möglichen Probleme zu vorhersehen und z.B. die Inputs auf jede Möglichkeit zu testen -- wie bei dem Volumen eines Quaders. Letztenendes ist es egal, welchen Datentyp die einzelnen Dimensionen haben, Hauptsache man kann sie mit Null vergleichen (Zwekcs Bestimmung, ob die Längen nicht negativ sind) und man kann sie miteinander multiplizieren.

Man kann in einem solchen Fall die Struktur:
```
try:
    # Anweisungen, die eigentlich an sich richtig sind, aber aufgrund von irgendwas doch mal schief gehen können
except (FehlerDenIchErwarte,KommaSepariert):
    # Anweisungen, die ausgeführt werden sollen, 
    # falls die ersten Anweisungen im vorherigen 
    # Block einen der nach except aufgeführten Fehlern 
    # verursacht haben
```
Zum Beispiel die letzte Aufgabe aus der Hausaufgabe kann so gelöst werden:

````python
def quaderVolumen(a,b,c):
    try:
        if a<0 or b<0 or c<0:
            return(-1)
        return(a*b*c)
        
    except (TypeError,ValueError): # Andere Fehler verursachen eine normale Fehlerreaktion!
        return(-1)

print(quaderVolumen(1,2,4))
print(quaderVolumen(-1,-2,-3))
print(quaderVolumen("Katze",0,"Hund"))
````

# Aufgabe in der Kursstunde:

- Eine Eingabe des Benutzers einlesen.
- Mache daraus eine Zahl und verdoplle sie.
- Wenn das nicht klappt schreib "das war keine Zahl"

Nutze dafür ein Try-Except Block.

# Musterlösung

````python
userEntry = input("Gib etwas ein. Wenn es eine Zahl ist, wird sie verdoppelt")
try:
    number = float(userEntry)
    print("Das doppelte von",number,"ist",number*2)
except ValueError:
    print("Das war wohl keine Zahl, die ich verdoppeln könnte")
````
