```python
print('Das ist eine Zeichenfolge. Da sie in Apostrophen eingefasst ist, kann ich die "normalen" Anführungszeichen einfach so verwenden.')
print("Das ist auch eine Zeichenfolge. Da sie in den doppelten Anführungszeichen eingefasst ist, kann ich hier den Apostroph verwenden: She's right!")
print("Ich kann aber einfach die \"speziellen\" Zeichen wie Apostroph oder Anführungszeichen ein Backslash \'\\\' voranstellen, dann funktionieren sie als \'normale\' Buchstaben.")
print("Es gibt auch einige spezielle Kombinationen mit dem Backslash: ein \\n macht eine neue\nZeile")
print(3) # ich kann einfach auch eine Zahl ausgeben - hier eine ganze Zahl
print(3.5) # und hier eine reelle Zahl

# Man kann auch mehrere verschiedene Sachen auf einmal ausgeben, sie werden mit einem Leerzeichen getrennt
print("Die Zahl pi (π) beträgt ungefähr",3.141592653589796368,"prüft aber genau, was ausgegeben wurde!")
```

```python
# Man kann auch mehrere verschiedene Sachen auf einmal ausgeben, sie werden mit einem Leerzeichen getrennt
print("Die Zahl pi (π) beträgt ungefähr",3.141592653589796368,"prüft aber genau, was ausgegeben wurde!")
print("wenn man dem Befehl 'print()' mehrere Argumente übergibt, werden sie einfach nacheinander ausgegeben, mit einem Leerzeichen dazwischen - siehe oben")
print("ich kann auch" + "zwei Zeichenketten \"addieren\". Das Ergebnis ist eine neue, längere Zeichenkette die aus den zwei \"zusammengeklebt\" wurde. OHNE Leerzeichen")
```

```python
print("3+4\t=",            3 + 4)      # Addition
print("\"3\"+\"4\"\t=",    "3" + "4")  # Zeichenketten können auch addiert werden - es entsteht dabei eine neue, längere Zeichenkette

print("3x4\t=",            3 * 4  )    # Multplikation zwei ganzen Zahlen ergibt eine ganze Zahl
print("3x4.0\t=",          3 * 4.0)    # 4 ist eine ganze Zahl, 4.0 ist aber eine reelle Zahl, das Ergebnis ist dann auch eine reelle Zahl
print("3x\"p\"\t=",        3 * "p")    # Zeichenketten können durch eine ganze Zahl multipliziert werden - die Zeichenkette wird einfach so oft wiederholt

print("3-4\t=",            3 - 4)      # Subtraktion

print("11/4\t=",           11 / 4)     # Division - "reelle"
print("11//4\t=",          11 //4)     # ganzzählige Division
print("11%4\t=",           11 % 4)     # Modulo == der Rest einer ganzzähligen Division

print("11**2\t=",          11**2)      # Potenzrechnung, ganzzählig
print("121**0.5=",       121**0.5)   # Wurzel ziehen - obwohl mathematisch gesehen √121 eine ganze Zahl ist, liefert diese Methode aber eine reelle Zahl als Ergebnis.
```

```python
print("Das Ergebnis der Rechnung:\n  [3 x (5 + 2)]/2 = \nist:",(3*(5+2))/2)
print("Achtung: auch wenn in der mathematischen Notation eckige [] und geschweifte {} Klammern verwendet werden, im Programm nutzt man nur die runden Klammern um die Reihenfolge der Berechnung zu bestimmen!")
```

`input()` gibt immer eine Zeichenkette zurück, wenn man eine Zahl braucht, muss man die Zeichenkette in eine Zahl wandeln:

```python
# float() macht, soweit möglich, aus einer Zeichenkette eine Zahl:

# Die Breite des Raumes:
breite = float(input("Wie breit ist der Raum?"))

# Die Länge des Raumes:
länge = float(input("Wie lang ist der Raum?"))

# int() macht, soweit möglich, aus einer Zeichenkette eine _ganze_ Zahl. 
personen = int(input("Wie viele Personen nutzen diesen Raum?"))

# str() macht, soweit möglich, aus dem Argument eine Zeichenkette
print("Die Fläche des Raumes ist: " + str(breite*länge) + "m², das macht "+str(breite*länge/personen)+"m² pro Person.")

# "soweit möglich" bedeutet, dass wenn man anstatt einer Zahl z.B. ein Wort eingibt, wird ein Fehler gemeldet und das Programm stoppt. Wie man mit diesem Problem umgehen kann lernen wir später.
```
