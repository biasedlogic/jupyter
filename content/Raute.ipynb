# Hausaufgabe
Eine Raute in der vorgegebenen Größe zeichnen, höchstens 10.

**Beispiel:** Eine Raute der Größe 5 (fünf „Etagen”) sieht so aus:
```
    *
   * *
  *   *
 *     *
*       *
 *     *
  *   *
   * *
    *
```
Denke daran, dass eine Zeichenkette multipliziert mit einer Zahl eine Wiederholung der Zeichenkette bedeutet!

# Schmale Raute

````python
n = int(input("Wie groß soll die Raute sein?: "))-1
print(n*" "+"*")
for i in range(1,n+1):
    print((n-i)*" "+"*"+(2*i-1)*" "+"*")
for i in range(n-1,0,-1):
    print((n-i)*" "+"*"+(2*i-1)*" "+"*")
if n>1:
    print(n*" "+"*")
````

# Flachere Raute

````python
n = int(input("Wie groß soll die Raute sein?: "))-1
print(2*n*" "+"*")
for i in range(1,n+1):
    print(2*(n-i)*" "+"*"+(4*i-1)*" "+"*")
for i in range(n-1,0,-1):
    print(2*(n-i)*" "+"*"+(4*i-1)*" "+"*")
if n>1:
    print(2*n*" "+"*")
````

# Raute mit wenig Formeln

````python
groesse = int(input('Wie gross soll die Raute sein?'))
# Obere Haelfte der Raute:
leerzeichenVorne = groesse-1
leerzeichenMitte = 1
print(' '*leerzeichenVorne+'*')

for n in range(groesse-1):
    leerzeichenVorne-=1
    print(' '*leerzeichenVorne+'*'
          +' '*leerzeichenMitte+'*')
    leerzeichenMitte+=2
#Untere Haelfte der Raute - schwieriger
leerzeichenMitte=groesse*2-5
leerzeichenVorne=0

for n in range(groesse-2):
    leerzeichenVorne+=1
    print(' '*leerzeichenVorne+'*'
          +' '*leerzeichenMitte+'*')
    leerzeichenMitte-=2
leerzeichenVorne+=1
print(' '*leerzeichenVorne+'*')
````
