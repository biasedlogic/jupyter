## Etwas in eine Liste einfügen...

```python
from random import randint
würfe = [] # das ist eine leere Liste
while True:
    w = randint(1,6)
    würfe.append(w)
    if w == 6:
        break
print("Endlich eine sechs!")

print("Bisher habe ich gewürfelt:",würfe,
     "Insgesamt sind das",sum(würfe),
     "Punkte in",len(würfe),"Würfen")
würfe2 = würfe.copy()
print(würfe)
würfe2.sort()
print(würfe2)
print(würfe)
```
