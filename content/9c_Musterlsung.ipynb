# Aufgabe 1 – Muss

Programmiere einen Küchentimer. Der Benutzer gibt die Zeit in Sekunden ein, der Timer soll dann die Sekunden herunterzählen und bei Null "FERTIG!" schreiben.

# Beispiel

Eingabe: `5`

Ausgabe:

```
5
4
3
2
1
FERTIG!
```

Dabei jede Zeile erscheint 1 Sekunde später, als die vorherige. 

Nutze dabei am Besten die Funktion `monotonic()` aus dem Modul `time` --> siehe dazu auch die Musterlösung der Aufgabe aus der vorletzten Kursstunde. Das Abmessen der Zeit haben wir in der letzten Stunde (12. Mai) ausgiebig besprochen.

## Musterlösung

````python
from time import monotonic, sleep
while True:
    try:
        zeit = int(input("Wie viele Sekunden soll der Timer abzählen?: "))
        break
    except ValueError:
        print("Bitte nur eine (ganze) Zahl eingeben!")

next = monotonic()
for i in range(zeit,0,-1):
    print(i)
    next+=1
    while next>monotonic():
        sleep(0.01)
print("FERTIG!")    
````

## Musterlösung - die Basisvariante

````python
# Musterlösung - lang[weilig]
from time import monotonic, sleep

def getTime():
    while True:
        timeStr = input("Wie lange soll der Timer laufen? hh:mm:ss oder mm:ss oder einfach ss ")
        try:
            t = timeStr.split(":")
            if len(t) == 3:
                h = int(t[0])
                m = int(t[1])
                s = int(t[2])
            elif len(t) == 2:
                h = 0
                m = int(t[0])
                s = int(t[1])
            elif len(t) == 1:
                h = m = 0
                s = int(t[0])
            else:
                raise ValueError
            while s>=60:
                s-=60
                m+=1
            while m>=60:
                m-=60
                h+=1
            return h,m,s
        except ValueError:
            print("Zeitformat falsch")

def countdown(h,m,s):
    next = monotonic()
    while h>0 or m>0 or s>0:
        print("Noch "+str(h)+"h, "+str(m)+"m, "+str(s)+"s")
        if s>0:
            s-=1
        else:
            s=59
            if m>0:
                m-=1
            else:
                m=59
                h-=1
        next+=1
        while next>monotonic():
            sleep(0.01)
    print("[--FERTIG!--]")

h,m,s = getTime()
countdown(h,m,s)
````

## Musterlösung -- Kompakt

````python
# Musterlösung - die kompakte Variante
from time import monotonic, sleep
from IPython.display import clear_output

def getTime():
    while True:
        timeStr = input("Wie lange soll der Timer laufen? hh:mm:ss oder mm:ss oder einfach ss ")
        try:
            t = [int(x) for x in timeStr.split(":")]
            t = [0]*(3-len(t)) + t
            for i in (2,1):
                while t[i]>59:
                    t[i-1]+=1
                    t[i]-=60
            return t
        except ValueError:
            print("Zeitformat falsch")

def countdown(h,m,s):
    s = (h*60+m)*60+s
    next = monotonic()
    while s>0:
        clear_output(wait=True)
        print(f"Noch {s//3600:02d}:{(s%3600)//60:02d}:{s%60:02d}")
        s-=1
        next+=1
        while next>monotonic(): sleep(0.01)
    clear_output(wait=True)
    print("[--FERTIG!--]")

h,m,s = getTime()
countdown(h,m,s)
````

## Hinweis 3

Der Befehl `in` prüft, ob etwas in einer Liste, oder in einer Zeichenkette vorkommt.  

````python
text = "abcdef"
if 'e' in text:
    print("'abcdef'.index('e') gibt",text.index('e'),"zurück.") 
else:
    print("'e' kommt in 'text' nicht vor")
if 'z' in text:
    print("'abcdef'.index('z') gibt",text.index('z'),"zurück.") 
else:
    print("'z' kommt in 'text' nicht vor")
````

## Musterlösung

````python
alphabet = "aäbcdefghijklmnoöpqrstuüvwxyz"
occurences = [0] * len(alphabet)
text = input("Gib den text, den man analysieren soll ein: ")

for c in text.lower(): 
    if c in alphabet: occurences[alphabet.index(c)]+=1

maximum = max(occurences)        
print("Am häufigsten, weil", maximum, "mal kommt vor: ",end="")
for i,n in enumerate(occurences):
    if n==maximum: print("'"+alphabet[i]+"' ",end="")
print()
````
