# Eine `while` Schleife

````python
antwort = "egal" #input("Bin ich richtig? Ja oder Nein:")

while antwort != "Ja" and antwort != "Nein": #  das != bedeutet ≠
    antwort = input("Bin ich richtig? Ja oder Nein:")
    
if antwort == "Ja":
    print("Hab' ich doch gesagt!")
else:
    print("Klar behauptest Du so was!")
````

# Bibliotheken: `import`...

````python
from random import randint # wir brauchen randint um Zufallszahlen zu generieren

w=0
while w!=6:            # so lange keine 6 gewürfelt wurde...
    w = randint(1,6)   #... weiter würfeln
    print(w,end=", ")  #... die gewürfelte Zahl ausgeben
print("Endlich eine sechs!")
````

````python
# Eine andere Lösung, die `break` verwendet, um die Endlosschleife zu unterbrechen
from random import randint

while True           # das endet nie...
    w=randint(1,6)
    print(w,end=", ")
    if w == 6:       
        break        # außer wir verwenden das Wort `break` um die Schleife zu beenden
print("Eine Sechs!")
````

# Eine `for ... in range():...` Schleife
![](wiederhole-n-mal.png)

````python
from random import randint # Wir brauchen `randint` um Zufallszahlen zu generieren

for i in range(5):      # range(n) generiert n Zahlen 0 bis n-1, die `for` Schleife dann "durchiteriert"
    print(randint(1,6)) # wir würfeln 5 Mal. Den Wert von i brauchen wir diesmal nicht.
````

````python
text = input("Gib mir ein Wort oder einen Satz:")
print("Das buchstabiere ich jetzt")
for c in text: # `for` kann auch andere Sets "durchiterieren" - Zeichenfolgen, Listen...
    print(c)
````

# Aufgabe 1: Einen Baum zeichnen

Ein Baum in der vorgegebenen Größe zeichnen, höchstens 10.

**Beispiel:** Ein Baum der Größe 5 (fünf „Etagen”) sieht so aus:
```
    *
   ***
  *****
 *******
*********
    |
```
Denke daran, dass eine Zeichenkette multipliziert mit einer Zahl eine Wiederholung der Zeichenkette bedeutet!

## Lösung

````python
size = int(input("Wie groß soll der Baum sein?"))
if size<1:
    print("Soll ich Dir nur die Wurzel zeichnen?! Hier ein Keimling, Größe 1:")
    size = 1
if size>10:
    print("Sorry, zu groß, wir machen max 10:")
    size = 10
for i in range(size):
    print(" "*(size-i) + "*"*(i*2+1))
print(" "*size + "|")
````
